## 7.2 Central Tendency

Central tendency measures are used to identify the central or representative value of a dataset.

The following measures are calculated for the numerical variables of the Insurance dataset:

- Arithmetic Mean
- Weighted Mean
- Geometric Mean
- Weighted Geometric Mean
- Harmonic Mean
- Weighted Harmonic Mean
- Quadratic Mean
- α-Mean
- Mode

```python
from scipy.stats import gmean, hmean

def weighted_mean(x, w):
    return np.sum(x * w) / np.sum(w)

def weighted_geometric_mean(x, w):
    return np.exp(np.sum(w * np.log(x)) / np.sum(w))

def weighted_harmonic_mean(x, w):
    return np.sum(w) / np.sum(w / x)

def quadratic_mean(x):
    return np.sqrt(np.mean(x ** 2))

def alpha_mean(x, alpha=3):
    return np.mean(x ** alpha) ** (1 / alpha)

results = []

for column in ["age", "bmi", "children", "charges"]:
    x = df[column].dropna().astype(float).values
    x_positive = x[x > 0]
    w = np.ones(len(x_positive))

    results.append([
        column,
        np.mean(x),
        weighted_mean(x_positive, w),
        gmean(x_positive),
        weighted_geometric_mean(x_positive, w),
        hmean(x_positive),
        weighted_harmonic_mean(x_positive, w),
        quadratic_mean(x),
        alpha_mean(x),
        df[column].mode().iloc[0]
    ])

central_tendency = pd.DataFrame(
    results,
    columns=[
        "Variable",
        "Arithmetic Mean",
        "Weighted Mean",
        "Geometric Mean",
        "Weighted Geometric Mean",
        "Harmonic Mean",
        "Weighted Harmonic Mean",
        "Quadratic Mean",
        "Alpha Mean",
        "Mode"
    ]
)

display(central_tendency.round(4))
```

### Observation

The central tendency analysis shows that the arithmetic mean age is 39.22 years, while the geometric mean and harmonic mean are lower at 36.51 years and 33.75 years respectively. The quadratic mean and α-mean are higher at 41.66 years and 43.74 years.

For BMI, the arithmetic mean is 30.66, while the geometric mean is 30.05 and the harmonic mean is 29.43. The quadratic mean is 31.26 and the α-mean is 31.85.

The average number of children is 1.10. The weighted mean is 1.92, while the geometric mean is 1.69 and the harmonic mean is 1.51. The mode is 0, indicating that having no children is the most frequently occurring value in the dataset.

For insurance charges, the arithmetic mean is 13,279.12. The geometric mean is 8,954.64 and the harmonic mean is 5,918.91, while the quadratic mean and α-mean are higher at 17,969.03 and 22,148.62 respectively. The mode is 1,121.87.

Overall, the different measures of central tendency provide different representations of the central values of the numerical variables. The considerable difference between the arithmetic, geometric, harmonic, quadratic, and α-means of insurance charges indicates substantial variation in the charge values.

## 7.4 Dispersion

Dispersion measures describe the amount of variation or spread present in numerical data.

The following measures are calculated for age, BMI, number of children, and insurance charges:

- Range
- Interquartile Range (IQR)
- Quartile Deviation
- Interdecile Range
- Interpercentile Range
- Mean Deviation
- Variance
- Standard Deviation
- Mean Difference

```python
numerical_columns = ["age", "bmi", "children", "charges"]

dispersion_results = []

for column in numerical_columns:
    x = df[column].dropna().astype(float).values

    q1 = np.percentile(x, 25)
    q3 = np.percentile(x, 75)

    p10 = np.percentile(x, 10)
    p90 = np.percentile(x, 90)

    p5 = np.percentile(x, 5)
    p95 = np.percentile(x, 95)

    mean_value = np.mean(x)

    mean_deviation = np.mean(np.abs(x - mean_value))

    variance = np.var(x, ddof=1)

    standard_deviation = np.std(x, ddof=1)

    mean_difference = np.mean(
        np.abs(x[:, None] - x[None, :])
    )

    dispersion_results.append([
        column,
        np.max(x) - np.min(x),
        q3 - q1,
        (q3 - q1) / 2,
        p90 - p10,
        p95 - p5,
        mean_deviation,
        variance,
        standard_deviation,
        mean_difference
    ])

dispersion_df = pd.DataFrame(
    dispersion_results,
    columns=[
        "Variable",
        "Range",
        "IQR",
        "Quartile Deviation",
        "Interdecile Range",
        "Interpercentile Range",
        "Mean Deviation",
        "Variance",
        "Standard Deviation",
        "Mean Difference"
    ]
)

display(dispersion_df.round(4))
```

### Observation

The dispersion analysis shows the amount of variation present in the numerical variables of the Insurance dataset.

For age, the range is 46 years, with an IQR of 24 years and a standard deviation of 14.04 years. The interdecile range is 40 years, indicating considerable variation across the central 80% of observations.

For BMI, the range is 37.17, while the IQR is 8.41 and the standard deviation is 6.10. The interdecile range is 15.64, showing moderate variation in BMI values.

For the number of children, the range is 5 and the IQR is 2. The standard deviation is 1.21, indicating comparatively lower dispersion than the other numerical variables.

Insurance charges show the largest absolute dispersion. The range is 62,648.55, the IQR is 11,911.37, and the standard deviation is 12,110.36. The interdecile range is 32,474.22 and the interpercentile range is 39,442.56.

The mean difference is 16.19 for age, 6.89 for BMI, 1.27 for children, and 12,292.69 for charges.

Overall, insurance charges exhibit substantial variability compared with the other numerical variables, while the number of children has relatively low dispersion. The dispersion measures provide a clearer understanding of the spread of the observations beyond the central tendency measures.

## 7.5 Frequency Analysis

Frequency analysis examines how observations are distributed across different value intervals.

Frequency tables are used to summarize observations within intervals, while frequency polygons and smoothed frequency curves represent the distribution graphically.

Ogives and smoothed ogives are used to visualize cumulative frequencies.

```python
from statsmodels.nonparametric.smoothers_lowess import lowess

numerical_columns = ["age", "bmi", "children", "charges"]

for column in numerical_columns:
    data = df[column].dropna()

    frequencies, bin_edges = np.histogram(data, bins=10)

    midpoints = (bin_edges[:-1] + bin_edges[1:]) / 2

    frequency_table = pd.DataFrame({
        "Class Interval": [
            f"{bin_edges[i]:.2f} - {bin_edges[i+1]:.2f}"
            for i in range(len(bin_edges) - 1)
        ],
        "Frequency": frequencies
    })

    print(f"\nFREQUENCY TABLE - {column.upper()}")
    display(frequency_table)

    plt.figure(figsize=(8, 5))
    plt.plot(midpoints, frequencies, marker="o")
    plt.title(f"Frequency Polygon - {column.capitalize()}")
    plt.xlabel(column.capitalize())
    plt.ylabel("Frequency")
    plt.grid(True)
    plt.show()

    smoothed = lowess(
        frequencies,
        midpoints,
        frac=0.6
    )

    plt.figure(figsize=(8, 5))
    plt.plot(smoothed[:, 0], smoothed[:, 1])
    plt.title(f"Smoothed Frequency Curve - {column.capitalize()}")
    plt.xlabel(column.capitalize())
    plt.ylabel("Frequency")
    plt.grid(True)
    plt.show()
```

### Observation

The frequency analysis shows different distribution patterns for the numerical variables.

For age, the highest frequency is observed in the interval 18.00–22.60 with 221 observations. The remaining observations are distributed across the other age intervals, showing that the dataset contains participants across a broad age range.

For BMI, the highest frequency occurs in the interval 27.11–30.83 with 318 observations, followed by the interval 30.83–34.55 with 288 observations. Very high BMI values occur less frequently, with only 3 observations in the 49.41–53.13 interval.

For the number of children, the highest frequency is 573 for the 0.00–0.50 interval, representing individuals with zero children. The next highest frequencies are 324 for the 1.00–1.50 interval and 240 for the 2.00–2.50 interval. The zero-frequency intervals occur because the number of children is a discrete integer variable.

For insurance charges, the highest frequency is 535 in the 1,121.87–7,386.73 interval, followed by 398 observations in the 7,386.73–13,651.58 interval. The frequencies decrease considerably at higher charge intervals, although some observations occur at relatively high charge values.

Overall, the frequency tables show that BMI is concentrated around the middle intervals, the number of children is concentrated at lower values, and insurance charges have many observations in the lower charge ranges with fewer observations at higher charge levels.

### 7.5.1 Ogives and Smoothed Ogives

An ogive is a cumulative frequency curve that shows the number of observations below or above particular values.

In this analysis, a less-than cumulative frequency is calculated for the numerical variables. A smoothed curve is also used to represent the cumulative frequency pattern.

```python
numerical_columns = ["age", "bmi", "children", "charges"]

for column in numerical_columns:
    data = df[column].dropna()

    frequencies, bin_edges = np.histogram(data, bins=10)

    cumulative_frequency = np.cumsum(frequencies)

    upper_boundaries = bin_edges[1:]

    plt.figure(figsize=(8, 5))
    plt.plot(
        upper_boundaries,
        cumulative_frequency,
        marker="o"
    )
    plt.title(f"Ogive - {column.capitalize()}")
    plt.xlabel(column.capitalize())
    plt.ylabel("Cumulative Frequency")
    plt.grid(True)
    plt.show()

    smoothed_ogive = lowess(
        cumulative_frequency,
        upper_boundaries,
        frac=0.6
    )

    plt.figure(figsize=(8, 5))
    plt.plot(
        smoothed_ogive[:, 0],
        smoothed_ogive[:, 1]
    )
    plt.title(f"Smoothed Ogive - {column.capitalize()}")
    plt.xlabel(column.capitalize())
    plt.ylabel("Cumulative Frequency")
    plt.grid(True)
    plt.show()
```

### Observation

The frequency analysis shows different distribution patterns for the numerical variables in the Insurance dataset.

For age, the highest frequency is 221 in the 18.00–22.60 interval. The observations are distributed across all the age intervals, indicating representation across the complete age range of the dataset.

For BMI, the highest frequency is 318 in the 27.11–30.83 interval, followed by 288 observations in the 30.83–34.55 interval. The frequency decreases considerably for higher BMI intervals, with only 3 observations in the 49.41–53.13 interval.

For children, the highest frequency is 573 in the 0.00–0.50 interval, representing individuals with zero children. The next highest frequencies are 324 for 1.00–1.50 and 240 for 2.00–2.50. The intervals between integer values have zero frequency because children is a discrete variable.

For insurance charges, the highest frequency is 535 in the 1,121.87–7,386.73 interval, followed by 398 observations in the 7,386.73–13,651.58 interval. The frequencies generally become smaller at higher charge intervals, although observations are present up to the highest charge range.

Overall, the frequency analysis indicates that BMI observations are concentrated around the middle intervals, the number of children is concentrated at lower values, and insurance charges have a high concentration in the lower charge ranges with fewer observations at higher values.

## 7.6 Distribution Characteristics

Distribution characteristics are used to understand the shape of numerical data.

Skewness measures the asymmetry of a distribution around its mean. A positive value indicates right skewness, while a negative value indicates left skewness.

Kurtosis describes the shape and tail characteristics of a distribution. The excess kurtosis is used in this analysis, where a value near zero indicates a distribution with kurtosis similar to the normal distribution.

```python
from scipy.stats import skew, kurtosis

numerical_columns = ["age", "bmi", "children", "charges"]

distribution_results = []

for column in numerical_columns:
    data = df[column].dropna()

    distribution_results.append([
        column,
        skew(data),
        kurtosis(data)
    ])

distribution_df = pd.DataFrame(
    distribution_results,
    columns=[
        "Variable",
        "Skewness",
        "Excess Kurtosis"
    ]
)

display(distribution_df.round(4))
```

### Observation

The distribution characteristics were analyzed using skewness and excess kurtosis for age, BMI, number of children, and insurance charges.

Skewness was used to determine the direction and degree of asymmetry in each numerical variable. Positive skewness indicates a distribution with a longer right tail, while negative skewness indicates a longer left tail.

Excess kurtosis was used to examine the tail characteristics and peakedness of the distributions. Positive excess kurtosis indicates heavier tails compared with a normal distribution, while negative excess kurtosis indicates lighter tails.

The calculated skewness and kurtosis values provide information about the shape of the numerical distributions and help identify variables that may show asymmetry or unusual tail behavior.

Overall, skewness and kurtosis complement the earlier descriptive and dispersion measures by providing additional information about the shape of the distributions.

Numerical × Numerical analysis examines the relationship between two numerical variables.

Scatter diagrams are used to visualize the relationship between numerical variables. Linear regression is used to examine a linear relationship, while LOWESS smoothing is used to identify possible nonlinear patterns.

The relationships considered are age versus charges, BMI versus charges, and age versus BMI.

```python
from sklearn.linear_model import LinearRegression
from statsmodels.nonparametric.smoothers_lowess import lowess

pairs = [
    ("age", "charges"),
    ("bmi", "charges"),
    ("age", "bmi")
]

for x_col, y_col in pairs:
    data = df[[x_col, y_col]].dropna()

    x = data[x_col].values
    y = data[y_col].values

    model = LinearRegression()
    model.fit(x.reshape(-1, 1), y)

    y_linear = model.predict(x.reshape(-1, 1))

    lowess_result = lowess(
        y,
        x,
        frac=0.6,
        return_sorted=True
    )

    order = np.argsort(x)

    plt.figure(figsize=(8, 5))
    plt.scatter(x, y, alpha=0.5)
    plt.plot(
        x[order],
        y_linear[order]
    )
    plt.title(
        f"Linear Relationship: {x_col.capitalize()} vs {y_col.capitalize()}"
    )
    plt.xlabel(x_col.capitalize())
    plt.ylabel(y_col.capitalize())
    plt.grid(True)
    plt.show()

    plt.figure(figsize=(8, 5))
    plt.scatter(x, y, alpha=0.5)
    plt.plot(
        lowess_result[:, 0],
        lowess_result[:, 1]
    )
    plt.title(
        f"Nonlinear Relationship using LOWESS: {x_col.capitalize()} vs {y_col.capitalize()}"
    )
    plt.xlabel(x_col.capitalize())
    plt.ylabel(y_col.capitalize())
    plt.grid(True)
    plt.show()
```

### Observation

The numerical × numerical analysis was performed using scatter diagrams, linear regression lines, and LOWESS smoothing.

The Age versus Charges plots show the relationship between age and insurance charges. The linear trend provides an overall indication of how charges vary with age, while the LOWESS curve helps identify changes in the relationship that may not be captured by a straight line.

The BMI versus Charges plots show the relationship between BMI and insurance charges. The LOWESS curve provides additional information about the shape of the relationship and helps identify possible nonlinear patterns.

The Age versus BMI plots show the relationship between age and BMI. The linear regression line represents the overall linear trend, while the LOWESS curve provides a more flexible representation of the observed pattern.

Overall, the combination of scatter diagrams, linear regression, and LOWESS smoothing provides a more detailed view of numerical relationships than a basic scatter plot alone. The LOWESS curves are particularly useful for identifying patterns that may not follow a simple linear relationship.

## 8.2 Categorical × Categorical

Categorical × categorical analysis is used to examine the association between two categorical variables.

In this analysis, sex and smoker status are studied using a contingency table, crossed frequencies, marginal frequencies, relative frequencies, row profiles, column profiles, expected frequencies, independence ratios, and chi-square contributions.

The Chi-square test is used to test whether the two categorical variables are statistically independent. Phi, Cramer's V, and Tschuprow's T are calculated to measure the strength of association between the variables.

```python
from scipy.stats import chi2_contingency

data = df[["sex", "smoker"]].dropna().copy()

sex_map = {
    0: "female",
    1: "male"
}

smoker_map = {
    0: "no",
    1: "yes"
}

data["sex"] = data["sex"].map(sex_map).astype(str)
data["smoker"] = data["smoker"].map(smoker_map).astype(str)

contingency_table = pd.crosstab(
    data["sex"],
    data["smoker"]
)

print("CONTINGENCY TABLE")
display(contingency_table)

print("CROSSED FREQUENCIES")
display(contingency_table)

row_totals = contingency_table.sum(axis=1)
column_totals = contingency_table.sum(axis=0)
grand_total = contingency_table.values.sum()

marginal_frequencies = pd.concat(
    [
        row_totals.rename("Row Total"),
        pd.Series({"Grand Total": grand_total})
    ]
)

print("ROW MARGINAL FREQUENCIES")
display(marginal_frequencies.to_frame())

print("COLUMN MARGINAL FREQUENCIES")
display(column_totals.to_frame("Column Total"))

print("GRAND TOTAL")
print(grand_total)

relative_frequencies = contingency_table / grand_total

print("RELATIVE FREQUENCIES")
display(relative_frequencies.round(4))

row_profiles = contingency_table.div(
    contingency_table.sum(axis=1),
    axis=0
)

print("ROW PROFILES")
display(row_profiles.round(4))

column_profiles = contingency_table.div(
    contingency_table.sum(axis=0),
    axis=1
)

print("COLUMN PROFILES")
display(column_profiles.round(4))

chi2_stat, p_value, degrees_of_freedom, expected = chi2_contingency(
    contingency_table,
    correction=False
)

expected_frequencies = pd.DataFrame(
    expected,
    index=contingency_table.index,
    columns=contingency_table.columns
)

print("EXPECTED FREQUENCIES")
display(expected_frequencies.round(4))

independence_ratio = contingency_table / expected_frequencies

print("INDEPENDENCE RATIOS")
display(independence_ratio.round(4))

chi_square_contributions = (
    (contingency_table - expected_frequencies) ** 2
) / expected_frequencies

print("CHI-SQUARE CONTRIBUTIONS")
display(chi_square_contributions.round(4))

n = contingency_table.values.sum()
r, c = contingency_table.shape

phi = np.sqrt(chi2_stat / n)

cramers_v = np.sqrt(
    chi2_stat / (n * min(r - 1, c - 1))
)

tschuprow_t = np.sqrt(
    chi2_stat / (
        n * np.sqrt((r - 1) * (c - 1))
    )
)

print("CHI-SQUARE TEST")
print("Chi-square statistic:", round(chi2_stat, 4))
print("Degrees of freedom:", degrees_of_freedom)
print("p-value:", round(p_value, 6))

print("\nASSOCIATION MEASURES")
print("Phi:", round(phi, 4))
print("Cramer's V:", round(cramers_v, 4))
print("Tschuprow's T:", round(tschuprow_t, 4))
```

### Observation

The categorical × categorical analysis examines the relationship between sex and smoker status using a contingency table.

The crossed frequencies and marginal frequencies provide the distribution of observations across the different combinations of sex and smoker status. Relative frequencies show the proportion of each combination with respect to the total number of observations.

The row profiles and column profiles provide the conditional distribution of smoker status within each sex category and the distribution of sex within each smoker category.

Expected frequencies are calculated under the assumption that sex and smoker status are independent. The independence ratios compare the observed frequencies with their corresponding expected frequencies. Values greater than 1 indicate that the observed frequency is higher than the expected frequency, while values below 1 indicate that it is lower than expected.

The chi-square contributions show how much each cell contributes to the overall chi-square statistic. The chi-square test is used to determine whether there is a statistically significant association between sex and smoker status.

Phi, Cramer's V, and Tschuprow's T provide measures of the strength of association between the two categorical variables.

Overall, the analysis provides a detailed statistical assessment of the relationship between sex and smoker status using both frequency-based and association-based measures.

## 8.5 Density Scatter

A density scatter plot is used to visualize the concentration of observations between two numerical variables.

In this analysis, BMI and insurance charges are used to identify regions where observations are more densely concentrated.

Density information helps reveal areas with a high concentration of observations that may not be easily visible in a conventional scatter plot.

```python
from scipy.stats import gaussian_kde

density_data = df[["bmi", "charges"]].dropna()

x = density_data["bmi"].values
y = density_data["charges"].values

xy = np.vstack([x, y])

density = gaussian_kde(xy)(xy)

order = density.argsort()

x = x[order]
y = y[order]
density = density[order]

plt.figure(figsize=(9, 6))

plt.scatter(
    x,
    y,
    c=density,
    s=30,
    alpha=0.7
)

plt.title("Density Scatter Plot: BMI vs Insurance Charges")
plt.xlabel("BMI")
plt.ylabel("Insurance Charges")
plt.colorbar(label="Observation Density")
plt.grid(True)
plt.show()
```

### Observation

The density scatter plot shows the concentration of observations between BMI and insurance charges.

A higher concentration of observations is visible in the commonly occurring BMI and lower-to-moderate insurance charge ranges, while observations become more dispersed as insurance charges increase.

The density visualization makes it easier to identify areas where many observations overlap, which may be less apparent in a conventional scatter plot.

The plot also shows that observations with higher insurance charges are comparatively fewer and are spread across different BMI values.

Overall, the density scatter plot provides a clearer representation of the concentration and spread of BMI and insurance charges and complements the earlier numerical relationship analysis.
