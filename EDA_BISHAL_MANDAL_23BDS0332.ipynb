## Step 1: Import Required Libraries

The required Python libraries are imported for data manipulation, numerical computations, visualization, and preprocessing.

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import LabelEncoder

import warnings
warnings.filterwarnings("ignore")

%matplotlib inline
```

```python
url = "https://raw.githubusercontent.com/salemprakash/EDA/main/Data/insurance.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully.")

df
```

The isnull().sum() function is used to identify the number of missing values present in each column of the dataset.

## Step 10: Check for Duplicate Values

The duplicated().sum() function is used to identify the total number of duplicate records present in the dataset.

## Step 11: Data Cleaning

Data cleaning is performed to improve the quality and consistency of the dataset. Duplicate records are removed, and the dataset is checked again to ensure that no duplicate values remain.

```python
df = df.drop_duplicates()

df.duplicated().sum()
```

## Step 12: Check Missing Values After Data Cleaning

After removing duplicate records, the dataset is checked again for missing values to ensure that the data is complete and ready for further analysis.

```python
df.isnull().sum()
```

## Step 13: Data Transformation

Data transformation is performed to convert categorical variables into numerical form. Label Encoding is used to transform the categorical columns sex, smoker, and region into numerical values for further analysis.

```python
le = LabelEncoder()

df["sex"] = le.fit_transform(df["sex"])
df["smoker"] = le.fit_transform(df["smoker"])
df["region"] = le.fit_transform(df["region"])

df.head()
```

## Step 14: Check the Transformed Data Types

After transforming the categorical variables into numerical values, the data types of all columns are checked to verify that the transformation has been performed successfully.

```python
df.dtypes
```

## Step 15: Display the Transformed Dataset

The transformed dataset is displayed to verify that the categorical variables have been successfully converted into numerical values.

```python
df.head()
```

Step 16: Univariate Analysis
Univariate analysis involves analyzing one variable at a time. It helps in understanding the distribution, central tendency, spread, and frequency of individual variables in the dataset. In this section, three different visualizations are used to explore the insurance data.

16.1 Distribution of Age
A histogram is used to visualize the distribution of the age variable.

```python
plt.figure(figsize=(8, 5))
sns.histplot(df["age"], bins=20, kde=True)
plt.title("Distribution of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

### 16.2 Distribution of BMI

A histogram is used to visualize the distribution of the bmi variable and understand how BMI values are spread across the dataset.

```python
plt.figure(figsize=(8, 5))
sns.histplot(df["bmi"], bins=20, kde=True)
plt.title("Distribution of BMI")
plt.xlabel("BMI")
plt.ylabel("Frequency")
plt.show()
```

16.3 Distribution of Insurance Charges
A histogram is used to visualize the distribution of the charges variable and understand how the insurance charges are distributed across the dataset.

```python
plt.figure(figsize=(8, 5))
sns.histplot(df["charges"], bins=20, kde=True)
plt.title("Distribution of Insurance Charges")
plt.xlabel("Insurance Charges")
plt.ylabel("Frequency")
plt.show()
```

16.4 Box Plot of Insurance Charges
A box plot is used to visualize the distribution of the charges variable. It helps identify the median, spread, and potential outliers in insurance charges.

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x=df["charges"])
plt.title("Box Plot of Insurance Charges")
plt.xlabel("Insurance Charges")
plt.show()
```

Step 17: Bivariate Analysis
Bivariate analysis is used to examine the relationship between two variables. It helps identify patterns, trends, and associations between different attributes of the insurance dataset.

### 17.1 Age vs Insurance Charges

A scatter plot is used to analyze the relationship between age and charges.

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x="age", y="charges", data=df)
plt.title("Age vs Insurance Charges")
plt.xlabel("Age")
plt.ylabel("Insurance Charges")
plt.show()
```

### 17.2 BMI vs Insurance Charges

A scatter plot is used to analyze the relationship between bmi and charges.

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x="bmi", y="charges", data=df)
plt.title("BMI vs Insurance Charges")
plt.xlabel("BMI")
plt.ylabel("Insurance Charges")
plt.show()
```

### 17.3 Smoker Status vs Insurance Charges

A box plot is used to compare insurance charges between different smoker status groups.

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x="smoker", y="charges", data=df)
plt.title("Smoker Status vs Insurance Charges")
plt.xlabel("Smoker Status")
plt.ylabel("Insurance Charges")
plt.show()
```

### 17.4 Children vs Insurance Charges

A box plot is used to examine the relationship between the number of children and insurance charges.

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x="children", y="charges", data=df)
plt.title("Number of Children vs Insurance Charges")
plt.xlabel("Number of Children")
plt.ylabel("Insurance Charges")
plt.show()
```

## Step 18: Multivariate Analysis

Multivariate analysis examines three or more variables simultaneously. It helps identify interactions and relationships among multiple variables in the insurance dataset.

### 18.1 Pair Plot of Numerical Variables

A pair plot is used to visualize pairwise relationships among the numerical variables in the dataset.

```python
sns.pairplot(df[["age", "bmi", "children", "charges"]])
plt.show()
```

### 18.2 Age, BMI and Insurance Charges by Smoker Status

A scatter plot with smoker status as the hue is used to analyze the combined relationship among age, BMI, and insurance charges.

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    x="age",
    y="charges",
    size="bmi",
    hue="smoker",
    data=df,
    sizes=(20, 200)
)
plt.title("Age, BMI and Insurance Charges by Smoker Status")
plt.xlabel("Age")
plt.ylabel("Insurance Charges")
plt.show()
```

### 18.3 Correlation Heatmap

A correlation heatmap is used to examine the strength and direction of relationships among the numerical variables.

```python
plt.figure(figsize=(10, 6))
sns.heatmap(df.corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap of Insurance Dataset")
plt.show()
```
